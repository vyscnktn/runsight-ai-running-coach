# Drop unneeded columns

In the previous step (`1_data_preprocess.ipynb`) the raw JSON data was
processed and saved as `endomondoHR_proper.csv`. The job of this notebook
is very simple but important: reopen the CSV and remove the columns that
add no value for modelling. The goal is for every later notebook to work
with fewer and more meaningful columns and a smaller file.

```python
import pandas as pd
import numpy as np
```

```python
PATH = "data/endomondoHR_proper.csv"
df = pd.read_csv(PATH)
```

## First, look at the current state

Before deciding what to delete, we inspect the columns and the first few
rows. Two things stand out:

- **`Unnamed: 0`** - this is not a column that pandas generates on its
  own. In the previous notebook `df.to_csv(...)` was called WITHOUT
  `index=False`, so the DataFrame's row index (0, 1, 2, ...) was also
  written to the CSV as a separate column. When the CSV is read back with
  `pd.read_csv`, pandas does not recognize this index column and reads it
  back as an ordinary data column named `Unnamed: 0`. It has no analytical
  value; it is just an accidental duplicate column.
- **`url`** - the link of each workout on the Endomondo website. For
  modelling (a classification based on numeric/time-series features such
  as speed, heart rate, location and duration) it carries no information
  and only takes up space.

```python
df.head(10)
```

```python
df.columns
```

## Remove both

The `df.columns` output confirms that the two suspected columns
(`Unnamed: 0`, `url`) are really there. We remove both with `drop`. We do
not touch the remaining columns (location, altitude, heart rate, speed,
time, user information); they will be used in the next steps (QC, feature
extraction).

```python
df = df.drop(columns=['Unnamed: 0','url'])
```

```python
df.head(10)
```

## Save the result - this time with `index=False`

We overwrite the same `endomondoHR_proper.csv` file, but this time with
the `index=False` parameter, so pandas does not write the row index to the
CSV as a separate column. This prevents the `Unnamed: 0` problem we saw at
the start from happening again: whoever reads this file from now on will
no longer run into an accidental index column.

```python
df.to_csv("data/endomondoHR_proper.csv", index=False)
```
